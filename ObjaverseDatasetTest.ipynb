```python
from ObjaverseDataset import ObjaverseDataset
from torch.utils.data import DataLoader
```

```python
dataset = ObjaverseDataset(category="duck", limit=10)
dataloader = DataLoader(dataset, batch_size=1, shuffle=False)
```

```python
for batch in dataloader:
    print("UID:", batch["uid"][0])
    print("Путь к .obj:", batch["obj_path"][0])
    print("Форма вершин (vertices):", batch["vertices"].shape)  # Ожидаем [1, N, 3]
    print("Форма граней (faces):", batch["faces"].shape)          # Ожидаем [1, M, 3]
    break
```

```python
sample = dataset[9]
mesh = sample["mesh"]

print(f"Показываем модель: {sample['uid']}")

# Интерактивный просмотр (откроется 3D-виджет, который можно крутить мышкой)
mesh.show()
```
