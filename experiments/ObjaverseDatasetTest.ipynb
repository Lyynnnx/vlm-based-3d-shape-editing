```python
from ObjaverseDataset import ObjaverseDataset
from torch.utils.data import DataLoader
```

Define dataloader

```python
dataset = ObjaverseDataset(uids=["41b9b4576fd043b9bcda02fc9e9fd138"] ,category = "car", start_index=1, limit=10)
dataloader = DataLoader(dataset, batch_size=1, shuffle=False, collate_fn=dataset.collate_fn)
```

Test data loading

```python
for batch in dataloader:
    print(batch.keys())
    print("UID:", batch["uid"][0])
    print(".obj path:", batch["obj_path"][0])
    print("vertices:", batch["vertices"][0].shape)
    print("faces:", batch["faces"][0].shape)
    break
```

Check the data sample in 3d

```python
sample = dataset[0]
mesh = sample["mesh"]

print(f"The model with id: {sample['uid']}")

mesh.show()
```
